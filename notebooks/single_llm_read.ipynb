# Single LLM Call for RE and Design

## Import Required Libraries

```python
from langchain_ollama import ChatOllama
from langchain_openai import ChatOpenAI
```

## Prompts for LLM API Calls

```python
req_design_message = [
    ("system", "You are an experienced requirements and design engineer. Help humans in analyzing software requirements and creating system designs and output the design in markdown format."),
    ("human", """Analyze the requirements and design the software for a tic-tac-toe game.""")
]
```

## LLM Call Using Local LLM

```python
llm = ChatOllama(model='gemma3:27b', temperature=0.2)
design = llm.invoke(req_design_message).content
```

```python
print (design)
```

## LLM Call Using Cloud LLM

```python
cloud_llm = ChatOpenAI(model='o4-mini')
cloud_design = cloud_llm.invoke(req_design_message).content
print(cloud_design)
```
